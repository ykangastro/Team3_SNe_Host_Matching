```python
import inspect
import textwrap
import pandas as pd
import astro_prost.helpers as prost_helpers

from astro_prost import associate_sample
from scipy.stats import gamma, uniform

# Temporary fix for the read-only DECaLS galaxy-size array.
# Run this cell again after restarting the kernel.
if not getattr(prost_helpers, "_shape_r_copy_patch_applied", False):
    source = textwrap.dedent(
        inspect.getsource(prost_helpers.calc_shape_props_decals)
    )

    old = 'candidate_hosts["shape_r"].values'
    new = 'candidate_hosts["shape_r"].to_numpy(copy=True)'

    if old not in source:
        raise RuntimeError(
            "Expected code not found in calc_shape_props_decals. "
            "The installed PROST version may differ."
        )

    source = source.replace(old, new)
    exec(
        compile(source, "<prost_notebook_patch>", "exec"),
        prost_helpers.__dict__,
    )
    prost_helpers._shape_r_copy_patch_applied = True

# Define a transient catalog.
transient_catalog = pd.DataFrame({
    "name": ["SN2025nou"],
    "ra": [331.058529],
    "dec": [17.0653057310261886],
})

# Define catalogs to search.
catalogs = ["glade", "decals", "panstarrs"]

# Define priors and likelihoods.
priors = {"offset": uniform(loc=0, scale=10)}
likes = {"offset": gamma(a=0.75)}

# Associate.
hosts = associate_sample(
    transient_catalog,
    priors=priors,
    likes=likes,
    catalogs=catalogs,
    name_col="name",
    coord_cols=("ra", "dec"),
    parallel=False,
    save=False,
)

display(hosts)
```
